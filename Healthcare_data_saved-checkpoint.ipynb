```python
# ── Real CPT codes with descriptions and typical cost ranges ──────────────
cpt_codes = [
    ("99213", "Office visit - established patient, low complexity",       80,  250),
    ("99214", "Office visit - established patient, moderate complexity",  120,  350),
    ("99232", "Subsequent hospital care, per day",                        100,  300),
    ("93306", "Echocardiography with doppler",                            800, 2500),
    ("71046", "Chest X-ray, 2 views",                                      60,  200),
    ("85025", "Complete blood count (CBC)",                                20,   80),
    ("80053", "Comprehensive metabolic panel",                             25,   90),
    ("99285", "Emergency dept visit - high complexity",                   600, 2000),
    ("27447", "Total knee replacement",                                  8000,25000),
    ("29881", "Knee arthroscopy with meniscectomy",                      2000, 7000),
    ("43239", "Upper GI endoscopy with biopsy",                          1000, 3500),
    ("99223", "Initial hospital care - high complexity",                  300,  900),
    ("36415", "Routine venipuncture",                                      10,   40),
    ("90837", "Psychotherapy, 60 minutes",                                100,  300),
    ("97110", "Therapeutic exercise",                                      50,  150),
    ("70553", "MRI brain with contrast",                                 1200, 4000),
    ("74177", "CT abdomen and pelvis with contrast",                     1500, 5000),
    ("99291", "Critical care, first 30-74 minutes",                      400, 1200),
    ("31622", "Bronchoscopy, diagnostic",                                1500, 5000),
    ("45378", "Colonoscopy, diagnostic",                                 1000, 3500),
]

# ── Real provider specialties ─────────────────────────────────────────────
specialties = [
    "Internal Medicine", "Family Practice", "Cardiology",
    "Orthopedic Surgery", "Emergency Medicine", "Gastroenterology",
    "Psychiatry", "Physical Therapy", "Radiology",
    "General Surgery", "Pulmonology", "Neurology"
]

# ── US States ─────────────────────────────────────────────────────────────
states = [
    "CA","TX","FL","NY","PA","IL","OH","GA","NC","MI",
    "NJ","VA","WA","AZ","MA","TN","IN","MD","MO","WI"
]

# ── ICD-10 diagnosis codes (common) ──────────────────────────────────────
icd10_codes = [
    ("I10",   "Essential hypertension"),
    ("E11.9", "Type 2 diabetes mellitus"),
    ("M54.5", "Low back pain"),
    ("J06.9", "Acute upper respiratory infection"),
    ("F32.1", "Major depressive disorder"),
    ("K21.0", "GERD with esophagitis"),
    ("M17.11","Primary osteoarthritis, right knee"),
    ("I25.10","Coronary artery disease"),
    ("J18.9", "Pneumonia, unspecified"),
    ("N39.0", "Urinary tract infection"),
]

# ── Generate 100,000 claims ───────────────────────────────────────────────
n = 100_000
records = []

for _ in range(n):
    cpt, desc, cost_min, cost_max = random.choice(cpt_codes)
    specialty                     = random.choice(specialties)
    state                         = random.choice(states)
    icd, diag                     = random.choice(icd10_codes)

    submitted   = round(random.uniform(cost_min, cost_max), 2)
    allowed_pct = random.uniform(0.30, 0.65)   # Medicare allows 30–65% of submitted
    allowed     = round(submitted * allowed_pct, 2)
    paid_pct    = random.uniform(0.75, 0.95)   # Medicare pays 75–95% of allowed
    paid        = round(allowed * paid_pct, 2)
    standardized = round(paid * random.uniform(0.90, 1.10), 2)

    benes    = random.randint(1, 500)
    services = int(benes * random.uniform(1.0, 3.5))

    records.append({
        "claim_id"          : fake.uuid4()[:8].upper(),
        "npi"               : fake.numerify(text="##########"),
        "provider_name"     : fake.last_name(),
        "specialty"         : specialty,
        "state"             : state,
        "cpt_code"          : cpt,
        "cpt_desc"          : desc,
        "icd10_code"        : icd,
        "diagnosis"         : diag,
        "place_of_service"  : random.choice(["Office","Outpatient Hospital",
                                              "Inpatient Hospital","Emergency Dept"]),
        "year"              : random.choice([2020, 2021, 2022, 2023]),
        "tot_benes"         : benes,
        "tot_srvcs"         : services,
        "avg_submitted_chrg": submitted,
        "avg_allowed_amt"   : allowed,
        "avg_paid_amt"      : paid,
        "avg_standardized_amt": standardized,
    })

df = pd.DataFrame(records)
print(f"Dataset created: {len(df):,} rows × {len(df.columns)} columns")
df.head()
```

```python
import duckdb

con = duckdb.connect()
con.register("medicare_claims", df)

total = con.execute("SELECT COUNT(*) FROM medicare_claims").fetchone()[0]
print(f"Loaded {total:,} rows into DuckDB — ready for SQL")
```

```python
q1 = """
SELECT
    cpt_code,
    cpt_desc                                        AS procedure_name,
    SUM(tot_srvcs)                                  AS total_services,
    ROUND(AVG(avg_paid_amt), 2)                     AS avg_payment,
    ROUND(SUM(tot_srvcs * avg_paid_amt), 0)         AS total_spend
FROM medicare_claims
GROUP BY cpt_code, cpt_desc
ORDER BY total_spend DESC
LIMIT 20
"""
top_procedures = con.execute(q1).df()
print("Query 1 done ✓")
top_procedures
```

```python
q2 = """
SELECT
    specialty,
    ROUND(AVG(avg_submitted_chrg), 2)                       AS avg_charged,
    ROUND(AVG(avg_allowed_amt), 2)                          AS avg_allowed,
    ROUND(AVG(avg_paid_amt), 2)                             AS avg_paid,
    ROUND(AVG(avg_paid_amt) /
          NULLIF(AVG(avg_submitted_chrg), 0) * 100, 1)      AS payment_to_charge_pct
FROM medicare_claims
GROUP BY specialty
ORDER BY avg_charged DESC
"""
specialty_payments = con.execute(q2).df()
print("Query 2 done ✓")
specialty_payments
```

```python
q3 = """
SELECT
    state,
    COUNT(DISTINCT npi)                             AS provider_count,
    SUM(tot_srvcs)                                  AS total_services,
    ROUND(AVG(avg_paid_amt), 2)                     AS avg_payment,
    ROUND(AVG(avg_submitted_chrg), 2)               AS avg_charged
FROM medicare_claims
WHERE cpt_code = '99213'
GROUP BY state
ORDER BY avg_payment DESC
"""
state_variation = con.execute(q3).df()
print("Query 3 done ✓")
state_variation
```

```python
q4 = """
SELECT
    year,
    cpt_code,
    cpt_desc                                        AS procedure_name,
    SUM(tot_srvcs)                                  AS total_services,
    ROUND(AVG(avg_paid_amt), 2)                     AS avg_payment,
    ROUND(SUM(tot_srvcs * avg_paid_amt), 0)         AS total_spend
FROM medicare_claims
GROUP BY year, cpt_code, cpt_desc
ORDER BY year, total_spend DESC
"""
yoy_trend = con.execute(q4).df()
print("Query 4 done ✓")
yoy_trend.head(20) 
```

```python
q5 = """
SELECT
    cpt_code,
    cpt_desc                                                AS procedure_name,
    specialty,
    tot_srvcs,
    tot_benes,
    ROUND(tot_srvcs / NULLIF(tot_benes, 0), 2)             AS services_per_beneficiary,
    RANK() OVER (
        PARTITION BY specialty
        ORDER BY tot_srvcs / NULLIF(tot_benes, 0) DESC
    )                                                       AS utilization_rank
FROM medicare_claims
QUALIFY utilization_rank <= 3
ORDER BY specialty, utilization_rank
"""
utilization = con.execute(q5).df()
print("Query 5 done ✓")
utilization
```

```python
# Payment to charge ratio
df['payment_to_charge_ratio'] = (
    df['avg_paid_amt'] / df['avg_submitted_chrg'].replace(0, pd.NA)
).round(4)

# Z-score anomaly flag
mean_r = df['payment_to_charge_ratio'].mean()
std_r  = df['payment_to_charge_ratio'].std()
df['is_anomaly'] = (df['payment_to_charge_ratio'] - mean_r).abs() > 2 * std_r

# Correlation
corr = df['tot_srvcs'].corr(df['avg_paid_amt'])

print("=== Payment Ratio Stats ===")
print(df['payment_to_charge_ratio'].describe().round(3))
print(f"\nAnomalies flagged       : {df['is_anomaly'].sum():,}")
print(f"Volume vs payment corr  : {corr:.4f}")

if corr < -0.1:
    print("Insight → Higher volume = LOWER avg payments")
elif corr > 0.1:
    print("Insight → Higher volume = HIGHER avg payments")
else:
    print("Insight → No strong relationship between volume and payment")
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 6))
top10 = top_procedures.head(10)

bars = ax.barh(
    top10['procedure_name'].str[:45],
    top10['total_spend'] / 1e9,
    color='steelblue', edgecolor='none'
)
ax.set_xlabel('Total Medicare Spend (Billions $)')
ax.set_title('Top 10 Procedures by Total Medicare Spend')
ax.invert_yaxis()

for bar, val in zip(bars, top10['total_spend'] / 1e9):
    ax.text(val + 0.05, bar.get_y() + bar.get_height() / 2,
            f'${val:.1f}B', va='center', fontsize=9)

plt.tight_layout()
plt.savefig('chart1_top_procedures.png', dpi=150, bbox_inches='tight')
plt.show()
print("Saved → chart1_top_procedures.png")
```
